# 🧹 Data Cleaning

## Objective
- Data cleaning is the process of identifying and correcting issues in a dataset, such as duplicate records, missing values, unnecessary columns, incorrect data types, unwanted spaces, and inconsistent data formats.
- We clean data to ensure that our analysis is accurate, reliable, and capable of solving business problems effectively without introducing errors or bias.
- Data cleaning is an important step because poor-quality data can lead to incorrect analysis, misleading insights, and poor business decisions.

```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\GitHub Repositories\Uber-Ride-Sharing-Analytics\datasets\uber.csv")
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df.columns
```

### Step 1: Removed Unnecessary Column

**Column:** `Unnamed: 0`

**Reason:**
The `Unnamed: 0` column did not provide any business or analytical value. It appeared to be an unnecessary index column and was removed to keep the dataset clean and focused on relevant features.

```python
has_duplicates = df.duplicated().any()
print(has_duplicates)
```

```python
duplicate_count = df.duplicated().sum()
print(f"There are {duplicate_count} duplicate rows.")
```

### Step 2: Checked Duplicate Records

**Result:**
No duplicate rows were found in the dataset.

**Reason:**
Duplicate records can lead to incorrect business insights, such as inflated trip counts and revenue. Since no duplicate rows were detected, no cleaning action was required.

```python
df_missing_rows = df[df.isna().any(axis=1)]
print(df_missing_rows)
```

```python
df.iloc[87946]
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
df.isna().sum()
```

### Step 3: Handled Missing Values

**Finding:**
One row contained missing values in the `dropoff_longitude` and `dropoff_latitude` columns.

**Decision:**
The affected row was removed from the dataset.

**Reason:**
Since only one row out of 200,000 contained missing geographic coordinates, removing it had a negligible impact on the dataset while preserving data quality. Filling these values could have introduced inaccurate location data and biased the analysis.

```python
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])
```

```python
df.info()
```

### Step 4: Fixed Incorrect Data Type

**Column:** `pickup_datetime`

**Issue:**
The `pickup_datetime` column was stored as an object (string) instead of a datetime data type.

**Action Taken:**
Converted the column to the datetime data type using `pd.to_datetime()`.

**Reason:**
Converting the column to datetime enables date and time-based analysis, such as extracting hours, weekdays, months, and identifying peak demand periods.

```python
df['fare_amount'].describe()
```

```python
zero_count = (df['fare_amount'] == 0).sum()
zero_count
```

```python
df = df[df['fare_amount'] >= 0]
```

```python
df['fare_amount'].min()
```

### Step 5: Handled Invalid Fare Values

**Finding:**
The `fare_amount` column contained negative values, which are not valid for completed Uber rides.

**Decision:**
Removed rows where `fare_amount` was less than 0.

**Reason:**
Negative fares are not possible in normal business operations and would distort revenue and fare analysis. Fare values equal to 0 were retained because, without cancellation or payment status information, they could represent valid business scenarios such as promotions or refunds.

```python
df['passenger_count'].describe()
```

```python
zero_count = (df['passenger_count'] == 0).sum()
zero_count
```

```python
six_count = (df['passenger_count'] > 6).sum()
six_count
```

```python
df = df[df['passenger_count'] <= 6]
```

```python
df['passenger_count'].max()
```

### Step 6: Handled Invalid Passenger Counts

**Finding:**
The `passenger_count` column contained one unrealistic value greater than 6.

**Decision:**
Removed rows where `passenger_count` was greater than 6.

**Reason:**
An Uber vehicle cannot realistically carry more than 6 passengers. Passenger counts equal to 0 were retained because the dataset lacked ride status or cancellation information, making it impossible to determine whether those records were invalid.

### Step 7: Handled Invalid Coordinates

**Finding:**
The dataset contained rows with latitude and longitude values outside the valid geographic ranges.

**Decision:**
Removed rows containing invalid coordinates.

**Reason:**
Latitude values must be between -90 and 90, and longitude values must be between -180 and 180. Invalid coordinates would lead to inaccurate geographic analysis and visualizations.
