# Feature Engineering

## Objective
Feature Engineering is the process of creating new columns (features) from existing data to make analysis easier and more meaningful.

It helps answer business questions, understand customer behavior, and generate valuable insights for stakeholders.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\GitHub Repositories\Uber-Ride-Sharing-Analytics\datasets\cleaned_uber.csv")
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'])
```

```python
df.info()
```

```python
df['pickup_hour'] = df['pickup_datetime'].dt.hour
df[['pickup_datetime', 'pickup_hour']].head()
```

### Step 1: Extract Pickup Hour

**Finding:**
The `pickup_datetime` column contains both date and time information.

**Feature Created:**
`pickup_hour`

**Reason:**
Extracting the pickup hour makes it easier to analyze hourly demand, identify peak hours, and answer business questions such as "At what hour do customers book the most Uber rides?"

```python
df['pickup_day'] = df['pickup_datetime'].dt.day
df[['pickup_datetime', 'pickup_day']].head()
```

### Step 2: Extract Pickup Day

**Feature Created:**
`pickup_day`

**Reason:**
The pickup day helps analyze ride patterns across different days of the month and supports time-based business analysis.

```python
df['pickup_month'] = df['pickup_datetime'].dt.month
df[['pickup_datetime', 'pickup_month']].head()
```

### Step 3: Extract Pickup Month

**Feature Created:**
`pickup_month`

**Reason:**
The pickup month helps analyze monthly ride demand, identify seasonal trends, and compare business performance across different months.

```python
df['pickup_year'] = df['pickup_datetime'].dt.year
df[['pickup_datetime', 'pickup_year']].head()
```

### Step 4: Extract Pickup Year

**Feature Created:**
`pickup_year`

**Reason:**
The pickup year helps analyze yearly ride trends, compare business performance across years, and identify long-term growth patterns.

```python
df['pickup_day_name'] = df['pickup_datetime'].dt.day_name()
df[['pickup_datetime', 'pickup_day_name']].head()
```

### Step 5: Extract Pickup Day Name

**Feature Created:**
`pickup_day_name`

**Reason:**
The pickup day name helps analyze ride demand by weekday, identify the busiest days, and support weekly business reporting.

```python
df['day_type'] = np.where(
    (df['pickup_day_name'] == 'Saturday') | (df['pickup_day_name'] == 'Sunday'),
    "Weekend",
    "Weekday"
)
df[['pickup_day_name', 'day_type']].head(10)
```

### Step 6: Create Day Type

**Feature Created:**
`day_type`

**Reason:**
This feature categorizes rides into **Weekday** and **Weekend**, making it easier to compare ride demand and revenue patterns between working days and weekends.
